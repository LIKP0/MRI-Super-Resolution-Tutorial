# 01. Data Preparation

## 1. About the Neurite-OASIS Data

[Neurite-OASIS](https://github.com/adalca/medical-datasets/blob/master/neurite-oasis.md) is a preprocessed version of the OASIS-1 brain MRI dataset. The volumes have already been registered, skull-stripped, bias-corrected, and standardized in shape and spacing.

In this notebook, we start from the Neurite-OASIS NIfTI files and prepare them for MRI super-resolution. The main steps are: intensity normalization, 2D slice resizing to **224 × 224**, low-resolution input generation from high-resolution images, and saving the processed data as `.npy` files for efficient training.

We use `axis=1` as the example slicing direction throughout this tutorial.

## 2. Setup

Set `src_dir` to your local Neurite-OASIS folder ([download here](https://surfer.nmr.mgh.harvard.edu/ftp/data/neurite/data/neurite-oasis.v1.0.tar)), and set `out_dir` to the folder where the processed `.npy` files will be saved. The helper functions used in this notebook are defined in `utils.py`.

Two important parameters are used throughout preprocessing:

- `slice_axis`: the axis used to extract 2D slices from each 3D MRI volume.
- `lr_scale`: the simulated super-resolution factor.

For example,`lr_scale=4` means that the model learns to recover a 1 mm³ high-resolution image from a 4 mm³ low-resolution input.

```python
from pathlib import Path
import random

import nibabel as nib
import numpy as np
from tqdm import tqdm
from torch.utils.data import DataLoader

from dataset import SRSliceDataset, TwoP5DSRSliceDataset, DegradeSRSliceDataset
from utils import (
    center_pad_or_crop_2d_plane,
    collect_oasis_pairs,
    describe_nifti,
    make_lr_from_hr,
    plot_batch,
    preprocess_oasis_pair,
    show_img_seg_pairs,
    subject_id_from_path,
)

# TODO: update src_dir to your local Neurite-OASIS folder.
src_dir = Path(r"E:\neurite-oasis.v1.0")

# Always write processed data to <project_root>/data.
cwd = Path.cwd().resolve()
project_root = cwd.parent if cwd.name == "tutorial" else cwd
out_dir = (project_root / "data").resolve()

target_size = 224
slice_axis = 1
lr_scale = 4
seed = 42

random.seed(seed)
np.random.seed(seed)
out_dir.mkdir(parents=True, exist_ok=True)

print("Source:", src_dir)
print("Output:", out_dir)
```

## 3. Understand the Input Data

Before preprocessing, it is useful to inspect the original files. A NIfTI file (`.nii.gz`) stores both the image array and metadata such as shape, voxel spacing, and orientation.

In this section, we collect paired `aligned_norm.nii.gz` and `aligned_seg4.nii.gz` files, print their basic properties, and visualize representative slices.

- `aligned_norm.nii.gz` is the anatomical MRI volume.
- `aligned_seg4.nii.gz` is a 4-region segmentation map that will later be used for region-wise evaluation.

```python
pairs = collect_oasis_pairs(src_dir)
print(f"Found {len(pairs)} image/segmentation pairs.")

for img_path, seg_path in pairs[:2]:
    print("Image:", img_path.relative_to(src_dir))
    print(describe_nifti(img_path))
    print("Seg:", seg_path.relative_to(src_dir))
    print(describe_nifti(seg_path))
    print()

sample_pairs = random.sample(pairs, k=min(2, len(pairs)))
show_img_seg_pairs(sample_pairs, axis=slice_axis)
```

## 4. Preprocess for Residual Super-Resolution

In a standard super-resolution setting, the model directly predicts the full high-resolution image from a low-resolution input. In this tutorial, we use a residual SR formulation: the low-resolution image is first upsampled to the high-resolution size, and the model learns to predict the missing high-frequency details, which is more friendly to models.

In this notebook, we treat the normalized Neurite-OASIS images as HR targets. LR inputs are simulated by downsampling each 2D slice plane and then upsampling it back to the HR size. This keeps the LR input and HR target spatially aligned.

The preprocessing pipeline includes:

- normalizing each MRI volume to `[0, 1]`
- padding or cropping the 2D slice plane to **224 × 224**
- generating LR inputs from HR images using downsampling and upsampling
- applying the same spatial processing to the segmentation masks
- saving `[LR, HR, mask]` as `.npy` files

We use `axis=1` as the slicing axis in this tutorial.


### 4.1 Normalize the HR Image

Normalize each MRI volume to `[0, 1]` so that all subjects share a consistent intensity range. This step makes the later training process more stable and easier to interpret.

```python
example_img_path, example_seg_path = pairs[0]

img = nib.load(str(example_img_path)).get_fdata(dtype=np.float32)
seg = nib.load(str(example_seg_path)).get_fdata(dtype=np.float32)

print("Before normalization:", float(img.min()), float(img.max()))

hr = (img - img.min()) / max(float(img.max() - img.min()), 1e-8)

print("After normalization:", float(hr.min()), float(hr.max()))
```

### 4.2 Pad or Crop the 2D Slice Plane

Next, resize the 2D slice plane to **224 × 224** using centered padding or cropping. The slicing axis (`axis=1`) is preserved, while the other two axes are adjusted to the target size.

```python
print("Before pad/crop:", hr.shape)

hr_224 = center_pad_or_crop_2d_plane(
    hr,
    target=target_size,
    axis=slice_axis,
    pad_value=0,
).astype(np.float32)

print("After pad/crop:", hr_224.shape)
```

### 4.3 Create LR from HR

Create the LR input by downsampling the HR image in the 2D slice plane and then upsampling it back to the HR shape. The upsampled LR image is used as the model input, and the original HR image is used as the target.


```python
lr = make_lr_from_hr(hr_224, scale=lr_scale, axis=slice_axis)

print("HR shape:", hr_224.shape, "range:", float(hr_224.min()), float(hr_224.max()))
print("LR shape:", lr.shape, "range:", float(lr.min()), float(lr.max()))
```

### 4.4 Align the Mask with the Processed Image

Apply the same padding/cropping operation to the segmentation map. This ensures that the mask labels remain spatially aligned with the processed LR and HR images.


```python
mask = seg.astype(np.float32)
mask_224 = center_pad_or_crop_2d_plane(
    mask,
    target=target_size,
    axis=slice_axis,
    pad_value=0,
).astype(np.float32)

print("Mask shape:", mask_224.shape)
print("Mask values:", np.unique(mask_224)[:10])
```

### 4.5 Save as NPY

Save the processed data as `.npy` files for faster loading with NumPy and PyTorch. Each saved file is channel-first.

With `slice_axis=1`, the saved array has shape `(3, 224, D, 224)`, where `D` is the number of slices along the slicing axis.

Channel order:

- `[0]`: LR input
- `[1]`: HR target
- `[2]`: segmentation mask


```python
example_data = np.stack([lr, hr_224, mask_224], axis=0).astype(np.float32)

print("Example NPY shape:", example_data.shape)
print("Channel 0 / LR:", example_data[0].shape)
print("Channel 1 / HR:", example_data[1].shape)
print("Channel 2 / mask:", example_data[2].shape)
```

### 4.6 Split the Dataset

Now apply the same preprocessing pipeline to all subjects and split them into train/validation/test sets with an **8:1:1** ratio. Each processed subject is saved as one `.npy` file under `out_dir/train`, `out_dir/val`, or `out_dir/test`.

Depending on your machine and the number of subjects, this step may take around 10 minutes and produce about 22 GB of `.npy` files.


```python
split_dirs = {
    "train": out_dir / "train",
    "val": out_dir / "val",
    "test": out_dir / "test",
}
for split_dir in split_dirs.values():
    split_dir.mkdir(parents=True, exist_ok=True)

shuffled_pairs = pairs.copy()
random.shuffle(shuffled_pairs)

n_total = len(shuffled_pairs)
n_train = int(0.8 * n_total)
n_val = int(0.1 * n_total)

split_pairs = {
    "train": shuffled_pairs[:n_train],
    "val": shuffled_pairs[n_train:n_train + n_val],
    "test": shuffled_pairs[n_train + n_val:],
}

saved_paths = {split: [] for split in split_pairs}

for split, current_pairs in split_pairs.items():
    print(f"{split}: {len(current_pairs)} subjects")

for split, current_pairs in split_pairs.items():
    for img_path, seg_path in tqdm(current_pairs, desc=f"Preprocessing {split}"):
        subject_id = subject_id_from_path(img_path)
        out_path = split_dirs[split] / f"{subject_id}.npy"

        data = preprocess_oasis_pair(
            img_path,
            seg_path,
            target=target_size,
            axis=slice_axis,
            scale=lr_scale,
        )
        np.save(out_path, data.astype(np.float16, copy=False))
        saved_paths[split].append(out_path)

print("Saved files:", {split: len(paths) for split, paths in saved_paths.items()})
saved_example_path = next(path for paths in saved_paths.values() for path in paths)
saved_example = np.load(saved_example_path, mmap_mode="r")
print("Example file:", saved_example_path)
print("Shape:", saved_example.shape)
print("Channel order: [LR input, HR target, mask]")
print("Dtype:", saved_example.dtype)
```

## 5. Load One Batch with `SRSliceDataset`

After preprocessing, we can load the saved `.npy` files with `SRSliceDataset`. This dataset extracts 2D slices along `slice_axis`, computes the residual target as `residual = dst - src` after any transforms, and returns one sample as a dictionary.

After batching with a `DataLoader`, each batch contains:

- `src`: LR tensor with shape `(B, 1, H, W)`
- `dst`: HR tensor with shape `(B, 1, H, W)`
- `residual`: HR-LR tensor with shape `(B, 1, H, W)`
- `mask`: segmentation tensor with shape `(B, 1, H, W)`
- `name`: sample identifiers, including the volume name and slice index


```python
train_dir = out_dir / "train"
npy_files = sorted(train_dir.glob("*.npy"))
print(f"Found {len(npy_files)} training npy files.")

if len(npy_files) == 0:
    raise RuntimeError(f"No .npy files found in {train_dir}. Run the preprocessing cell first, or set out_dir to the downloaded data.")

dataset = SRSliceDataset(
    filepaths=[str(p) for p in npy_files],
    slice_axis=slice_axis,
    slice_nums=32,
    cache_mode="memmap",
)
loader = DataLoader(dataset, batch_size=4, shuffle=True, num_workers=0)
batch = next(iter(loader))

print("src:", tuple(batch["src"].shape), batch["src"].dtype)
print("dst:", tuple(batch["dst"].shape), batch["dst"].dtype)
print("residual:", tuple(batch["residual"].shape), batch["residual"].dtype)
print("mask:", tuple(batch["mask"].shape), batch["mask"].dtype)
print("names:", batch["name"])
```

```python
plot_batch(batch, max_items=4)
```

## 6. Load One Batch with `TwoP5DSRSliceDataset`

The previous dataset uses a pure 2D setting: one LR slice is used to predict the corresponding HR slice. A **2.5D** setting still predicts a 2D target slice, but it provides neighboring slices as additional input channels. This gives the model more local 3D context while remaining much lighter than a full 3D model.

In `TwoP5DSRSliceDataset`, `src` contains `depth` neighboring LR slices. The `dst`, `residual`, and `mask` entries correspond to the center slice. The residual target is computed as `dst - center_src`.

You can observe the 2.5D data shape with the cell below.


```python
two_p5d_dataset = TwoP5DSRSliceDataset(
    filepaths=[str(p) for p in npy_files],
    slice_axis=slice_axis,
    slice_nums=32,
    depth=5,
    cache_mode="memmap",
)
two_p5d_loader = DataLoader(two_p5d_dataset, batch_size=4, shuffle=True, num_workers=0)
two_p5d_batch = next(iter(two_p5d_loader))

print("src:", tuple(two_p5d_batch["src"].shape), two_p5d_batch["src"].dtype)
print("dst:", tuple(two_p5d_batch["dst"].shape), two_p5d_batch["dst"].dtype)
print("residual:", tuple(two_p5d_batch["residual"].shape), two_p5d_batch["residual"].dtype)
print("mask:", tuple(two_p5d_batch["mask"].shape), two_p5d_batch["mask"].dtype)
print("names:", two_p5d_batch["name"])
```

## 7. More Realistic Degradation with `DegradeSRSliceDataset`

In the previous examples, we simplified the LR image as the result of downsampling followed by upsampling. This setting is useful for teaching and rapid validation, but it only covers part of the information loss involved in real MRI acquisition.

In practice, MRI degradation is usually caused by multiple coupled factors. Different scanning protocols and magnetic field strengths may introduce variations in spatial resolution and image contrast. Coil sensitivity and field inhomogeneity can lead to low-frequency intensity fluctuations, often referred to as bias field. The acquisition and reconstruction process may introduce point-spread effects and blurring. Noise, often approximated by Gaussian- or Rician-like distributions, can further corrupt fine details. In addition, post-processing and interpolation may alter image texture statistics. Therefore, MRI super-resolution is not simply about recovering frequencies lost during downsampling. Instead, the goal is to restore images that are structurally consistent, intensity-preserving, and useful for downstream analysis under mixed degradation conditions.

To better approximate this setting, `DegradeSRSliceDataset` applies a step-by-step degradation pipeline:

| Step | Operation | Parameters in `DegradeSRSliceDataset` |
|---|---|---|
| 1 | Clean HR image | Input HR stack (`dst`) |
| 2 | Bias field | `bias_strength_range=(0.10, 0.20)`, `bias_field_sigma_ratio=0.18` |
| 3 | Blur | `blur_sigma_range=(0.6, 1.4)` |
| 4 | Downsampling | `downsample_scale=4` |
| 5 | Noise | `noise_sigma_range=(0.0, 0.02)` |
| 6 | Upsampling back to HR size | Cubic interpolation (`scipy.ndimage.zoom`, `order=3`) |

For the `train` split, the degradation parameters are randomly sampled from the specified ranges. This introduces degradation diversity during training and encourages the model to learn a more robust SR mapping.

For the `val` and `test` splits, the midpoint of each parameter range is used as a fixed degradation setting. For example, `blur_sigma` is set to `(0.6 + 1.4) / 2 = 1.0`. This makes validation and testing deterministic, so that different models can be compared under the same degradation condition.

```python
degrade_dataset = DegradeSRSliceDataset(
    filepaths=[str(p) for p in npy_files],
    slice_axis=slice_axis,
    slice_nums=32,
    depth=9,
    cache_mode="memmap",
    split="train",
    base_seed=seed,
    bias_strength_range=(0.10, 0.20),
    blur_sigma_range=(0.6, 1.4),
    downsample_scale=4,
    noise_sigma_range=(0.0, 0.02),
    bias_field_sigma_ratio=0.18,
)

degrade_loader = DataLoader(degrade_dataset, batch_size=6, shuffle=True, num_workers=0)
degrade_batch = next(iter(degrade_loader))

print("src:", tuple(degrade_batch["src"].shape), degrade_batch["src"].dtype)
print("dst:", tuple(degrade_batch["dst"].shape), degrade_batch["dst"].dtype)
print("residual:", tuple(degrade_batch["residual"].shape), degrade_batch["residual"].dtype)
print("mask:", tuple(degrade_batch["mask"].shape), degrade_batch["mask"].dtype)
print("names:", degrade_batch["name"])
```

```python
plot_batch(degrade_batch, max_items=6)
```
