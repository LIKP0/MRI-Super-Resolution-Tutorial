# 00. Environment Setup

This notebook helps you set up the environment for the MRI super-resolution tutorial. You will create a Conda environment, install PyTorch and the required packages, register the environment as a Jupyter kernel, and run a few quick checks.

Please run installation commands in a terminal, not inside notebook cells, unless a cell explicitly asks you to do so.

## 1. Create a Conda Environment

Open a terminal, create a new Conda environment, and activate it.

Keep this environment active for all installation steps in this notebook. After activation, your terminal prompt should show `(mri_sr)`.
```bash
conda create -n mri_sr python=3.10 -y
conda activate mri_sr
```


Next, install `ipykernel` so that Jupyter can use this Conda environment as a notebook kernel.

Run the following commands in the same terminal after activating `mri_sr`.
```bash
pip install ipykernel
python -m ipykernel install --user --name mri_sr --display-name "Python (mri_sr)"
```


Then open or restart Jupyter and select the `Python (mri_sr)` kernel for this notebook.

If the kernel does not appear immediately, restart Jupyter, VS Code, or PyCharm and check the kernel list again.

## 2. Install PyTorch

Install PyTorch before installing the remaining project dependencies.

If you plan to use an NVIDIA GPU, first check whether your system can detect the GPU and driver:

```bash
nvidia-smi
```

If the command runs successfully, your system has access to the NVIDIA driver. The reported `CUDA Version` indicates the maximum CUDA runtime supported by the driver; it does not necessarily mean that a full CUDA Toolkit is installed.

If the command fails, common causes include: no NVIDIA GPU, a missing or incompatible driver, or `nvidia-smi` not being available in `PATH`.

Open the official PyTorch installation selector and copy the command that matches your operating system, package manager, and GPU/CPU setup:

https://pytorch.org/get-started/locally/

Run the selected command in the terminal where `(mri_sr)` is active. Complete this step before installing packages from `requirements.txt`.

Example command (Windows + NVIDIA GeForce RTX 5090 + CUDA 13.0):
```bash
pip3 install torch torchvision --index-url https://download.pytorch.org/whl/cu130
```
If you do not have a working NVIDIA setup, install the CPU version instead:
```bash
pip3 install torch torchvision
```
You can still run all later notebooks in CPU mode, but training/testing will be much slower.

Use the PyTorch selector for your own OS/GPU instead of copying the example directly.

## 4. Check PyTorch and CUDA

Run the cell below to confirm whether PyTorch can access CUDA.


````python
import torch

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")

if torch.cuda.is_available():
    print(f"CUDA version: {torch.version.cuda}")
    print(f"GPU count: {torch.cuda.device_count()}")
    print(f"GPU name: {torch.cuda.get_device_name(0)}")
else:
    print("Running on CPU.")
````

## 5. Check Required Packages

Run the cell below to confirm that the main packages import correctly.


````python
import importlib
from importlib.metadata import PackageNotFoundError, version

checks = {
    "diffusers": "diffusers",
    "pytorch_lightning": "pytorch-lightning",
    "nibabel": "nibabel",
    "torchmetrics": "torchmetrics",
    "numpy": "numpy",
    "scipy": "scipy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "tqdm": "tqdm",
    "yaml": "PyYAML",
    "ipywidgets": "ipywidgets",
}

failed = []

for module_name, package_name in checks.items():
    try:
        importlib.import_module(module_name)
        try:
            package_version = version(package_name)
        except PackageNotFoundError:
            package_version = "version not found"
        print(f"[OK] {module_name} ({package_version})")
    except Exception as exc:
        failed.append(module_name)
        print(f"[FAIL] {module_name}: {exc}")

if failed:
    print("\nFailed packages:", ", ".join(failed))
else:
    print("\nAll checked packages are available.")
````
